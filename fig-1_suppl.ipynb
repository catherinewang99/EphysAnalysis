```python
all_l_perf_naive, all_r_perf_naive = [],[]
all_early_lick_naive = []
all_ignore_naive = []
for paths in naive_paths:
    el = []
    ignore = []
    l_perf = []
    r_perf = []
    for path in paths:
        s1 = Session(path, passive=False, filter_low_perf=False, filter_by_stim=True)
        earlylick, correctarr, numtrials = s1.get_acc_conv(window=5)
        el += [np.mean(s1.early_lick)]
        ignore += [np.mean(s1.R_ignore + s1.L_ignore)]
        r, l, _ = s1.performance_in_trials(range(s1.num_trials))
        r_perf += [r]
        l_perf += [l]
    all_early_lick_naive.append(el)
    all_ignore_naive.append(ignore)
    all_l_perf_naive.append(l_perf)
    all_r_perf_naive.append(r_perf)


```

```python
all_l_perf_trained, all_r_perf_trained = [],[]
all_early_lick_trained = []
all_ignore_trained = []
for paths in trained_paths_high_performance:
    el = []
    ignore = []
    l_perf = []
    r_perf = []
    for path in paths:
        s1 = Session(path, passive=False, filter_low_perf=True, filter_by_stim=True)
        earlylick, correctarr, numtrials = s1.get_acc_conv(window=5)
        el += [np.mean(s1.early_lick)]
        ignore += [np.mean(s1.R_ignore + s1.L_ignore)]
        r, l, _ = s1.performance_in_trials(s1.i_good_non_stim_trials)
        r_perf += [r]
        l_perf += [l]
    all_early_lick_trained.append(el)
    all_ignore_trained.append(ignore)
    all_l_perf_trained.append(l_perf)
    all_r_perf_trained.append(r_perf)
```

```python


# Helper for marker cycling
markers = ['o', 's', 'D', '^', 'v', 'h', '>', '<', '*']
bar_width = 0.35
jitter = 0.08
x = np.arange(2)  # 0: naive, 1: trained

fig, ax = plt.subplots(figsize=(6,5))

# Flatten all L/R results, do not average within mouse
def flatten_full_list(lst):
    arr = []
    for x in lst:
        if isinstance(x, (list, np.ndarray)):
            arr.extend(x)
        else:
            arr.append(x)
    return np.array(arr)

l_naive = flatten_full_list(all_l_perf_naive)
r_naive = flatten_full_list(all_r_perf_naive)
l_trained = flatten_full_list(all_l_perf_trained)
r_trained = flatten_full_list(all_r_perf_trained)

mean_l_naive = np.mean(l_naive) if len(l_naive) > 0 else np.nan
mean_r_naive = np.mean(r_naive) if len(r_naive) > 0 else np.nan
mean_l_trained = np.mean(l_trained) if len(l_trained) > 0 else np.nan
mean_r_trained = np.mean(r_trained) if len(r_trained) > 0 else np.nan

# Plot bars of group means
ax.bar(x[0] - bar_width/2, mean_l_naive, bar_width, label='Left performance', color='red', alpha=0.7)
ax.bar(x[0] + bar_width/2, mean_r_naive, bar_width, label='Right performance', color='blue', alpha=0.7)
ax.bar(x[1] - bar_width/2, mean_l_trained, bar_width, color='red', alpha=0.7)
ax.bar(x[1] + bar_width/2, mean_r_trained, bar_width, color='blue', alpha=0.7)

# Use different marker for each mouse, assume each element in all_l_perf_* is a mouse, with multiple sessions/data points per mouse
def marker_indices(length, n_markers):
    # Repeats the marker cycle for required number of points
    reps = (length + n_markers - 1) // n_markers
    return [i % n_markers for i in range(length)]

# Expand lists so we know which mouse each point belongs to
def mouse_indices_from_nested(lst):
    idx = []
    for mousei, val in enumerate(lst):
        n = len(val) if isinstance(val, (list, np.ndarray)) else 1
        idx.extend([mousei]*n)
    return np.array(idx)

# For naive group
l_naive_points = flatten_full_list(all_l_perf_naive)
r_naive_points = flatten_full_list(all_r_perf_naive)
l_naive_mouse_idx = mouse_indices_from_nested(all_l_perf_naive)
r_naive_mouse_idx = mouse_indices_from_nested(all_r_perf_naive)

# For trained group
l_trained_points = flatten_full_list(all_l_perf_trained)
r_trained_points = flatten_full_list(all_r_perf_trained)
l_trained_mouse_idx = mouse_indices_from_nested(all_l_perf_trained)
r_trained_mouse_idx = mouse_indices_from_nested(all_r_perf_trained)

# Plot naive group with markers per mouse
for i in range(len(l_naive_points)):
    m = markers[l_naive_mouse_idx[i] % len(markers)]
    lx = x[0] - bar_width/2 + jitter
    ax.plot(lx, l_naive_points[i], marker=m, color='red', linestyle='none', markersize=8, zorder=2)
for i in range(len(r_naive_points)):
    m = markers[r_naive_mouse_idx[i] % len(markers)]
    rx = x[0] + bar_width/2 - jitter
    ax.plot(rx, r_naive_points[i], marker=m, color='blue', linestyle='none', markersize=8, zorder=2)
    # Connecting lines between left and right points for the same mouse if indices match
for mousei in set(l_naive_mouse_idx):
    lpoints = np.where(l_naive_mouse_idx == mousei)[0]
    rpoints = np.where(r_naive_mouse_idx == mousei)[0]
    # Connect each L/R pair per session (not just the mean)
    # Try pair-wise by order, fallback to min(len(l), len(r)) if unequal session count
    n_pairs = min(len(lpoints), len(rpoints))
    for j in range(n_pairs):
        lx = x[0] - bar_width/2 + jitter
        rx = x[0] + bar_width/2 - jitter
        ly = l_naive_points[lpoints[j]]
        ry = r_naive_points[rpoints[j]]
        ax.plot([lx, rx], [ly, ry], color='grey', linewidth=1, alpha=0.7, zorder=1)

# Plot trained group with markers per mouse
for i in range(len(l_trained_points)):
    m = markers[l_trained_mouse_idx[i] % len(markers)]
    lx = x[1] - bar_width/2 + jitter
    ax.plot(lx, l_trained_points[i], marker=m, color='red', linestyle='none', markersize=8, zorder=2)
for i in range(len(r_trained_points)):
    m = markers[r_trained_mouse_idx[i] % len(markers)]
    rx = x[1] + bar_width/2 - jitter
    ax.plot(rx, r_trained_points[i], marker=m, color='blue', linestyle='none', markersize=8, zorder=2)
for mousei in set(l_trained_mouse_idx):
    lpoints = np.where(l_trained_mouse_idx == mousei)[0]
    rpoints = np.where(r_trained_mouse_idx == mousei)[0]
    n_pairs = min(len(lpoints), len(rpoints))
    for j in range(n_pairs):
        lx = x[1] - bar_width/2 + jitter
        rx = x[1] + bar_width/2 - jitter
        ly = l_trained_points[lpoints[j]]
        ry = r_trained_points[rpoints[j]]
        ax.plot([lx, rx], [ly, ry], color='grey', linewidth=1, alpha=0.7, zorder=1)

# Dotted lines for thresholds
ax.axhline(0.5, color='k', linestyle='dotted', linewidth=1)
ax.axhline(0.7, color='gold', linestyle='dotted', linewidth=1)
ax.set_ylim(bottom=0.3, top=0.95)
ax.set_xticks(x)
ax.set_xticklabels(['Naive', 'Trained'])
ax.set_ylabel('Performance')
ax.set_title('Left and Right Performance in Naive and Trained Mice')
ax.legend()
plt.tight_layout()
plt.savefig(r'P:\paperfigures\fig1\left_right_performance_naive_vs_trained.pdf')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(6,5), sharey=True)

def plot_perf(ax, control_left, stim_left, control_right, stim_right, color_left, color_right, title):
    # Plot individual sessions as thin traces and points
    # Each argument is a list of lists (mice), with elements per mouse as list of sessions

    # The control columns come from *_trained arrays
    # Stim columns come from corresponding left/right_stim_* arrays

    n_mice = max(len(control_left), len(control_right), len(stim_left), len(stim_right))
    for i in range(n_mice):
        ctrl_l = np.array(control_left[i]) if i < len(control_left) else np.array([])
        ctrl_r = np.array(control_right[i]) if i < len(control_right) else np.array([])
        stim_l = np.array(stim_left[i]) if i < len(stim_left) else np.array([])
        stim_r = np.array(stim_right[i]) if i < len(stim_right) else np.array([])
        
        minlen_l = min(len(ctrl_l), len(stim_l))
        minlen_r = min(len(ctrl_r), len(stim_r))
        # For left trials
        for j in range(minlen_l):
            ax.plot([0, 1], [ctrl_l[j], stim_l[j]], color='red', linewidth=0.5, alpha=0.5)
            # ax.scatter([0], [ctrl_l[j]], color=color_left, s=16, alpha=0.7)
            # ax.scatter([1], [stim_l[j]], color=color_left, s=16, alpha=0.7)
        # For right trials
        for j in range(minlen_r):
            ax.plot([0, 1], [ctrl_r[j], stim_r[j]], color='blue', linewidth=0.5, alpha=0.5)
            # ax.scatter([0], [ctrl_r[j]], color=color_right, s=16, alpha=0.7)
            # ax.scatter([1], [stim_r[j]], color=color_right, s=16, alpha=0.7)

    # Flatten all values for average
    all_ctrl_l = np.array([val for sublist in control_left for val in sublist])
    all_stim_l = np.array([val for sublist in stim_left for val in sublist])
    all_ctrl_r = np.array([val for sublist in control_right for val in sublist])
    all_stim_r = np.array([val for sublist in stim_right for val in sublist])

    # Means and SEMs
    mean_ctrl_l = np.nanmean(all_ctrl_l) if len(all_ctrl_l) else np.nan
    sem_ctrl_l = np.nanstd(all_ctrl_l)/np.sqrt(np.sum(~np.isnan(all_ctrl_l))) if len(all_ctrl_l) else np.nan
    mean_ctrl_r = np.nanmean(all_ctrl_r) if len(all_ctrl_r) else np.nan
    sem_ctrl_r = np.nanstd(all_ctrl_r)/np.sqrt(np.sum(~np.isnan(all_ctrl_r))) if len(all_ctrl_r) else np.nan

    mean_stim_l = np.nanmean(all_stim_l) if len(all_stim_l) else np.nan
    sem_stim_l = np.nanstd(all_stim_l)/np.sqrt(np.sum(~np.isnan(all_stim_l))) if len(all_stim_l) else np.nan
    mean_stim_r = np.nanmean(all_stim_r) if len(all_stim_r) else np.nan
    sem_stim_r = np.nanstd(all_stim_r)/np.sqrt(np.sum(~np.isnan(all_stim_r))) if len(all_stim_r) else np.nan

    # Plot line between left and right
    # Lick left, control to stim (red)
    ax.plot([0, 1], [mean_ctrl_l, mean_stim_l], color=color_left, marker='o', markersize=13, linewidth=3, zorder=10)
    ax.errorbar([0], [mean_ctrl_l], yerr=sem_ctrl_l, color=color_left, fmt='o', markersize=13, linewidth=3, capsize=5, zorder=11)
    ax.errorbar([1], [mean_stim_l], yerr=sem_stim_l, color=color_left, fmt='o', markersize=13, linewidth=3, capsize=5, zorder=11)

    # Lick right, control to stim (blue)
    ax.plot([0, 1], [mean_ctrl_r, mean_stim_r], color=color_right, marker='o', markersize=13, linewidth=3, zorder=10)
    ax.errorbar([0], [mean_ctrl_r], yerr=sem_ctrl_r, color=color_right, fmt='o', markersize=13, linewidth=3, capsize=5, zorder=11)
    ax.errorbar([1], [mean_stim_r], yerr=sem_stim_r, color=color_right, fmt='o', markersize=13, linewidth=3, capsize=5, zorder=11)

    ax.set_xticks([0, 1])
    ax.set_xticklabels(['Control', 'Stim'])
    ax.set_title(title)
    ax.set_ylabel('Performance')
    ax.set_ylim(0.3, 0.9)
    ax.grid(True, axis='y', linestyle='--', alpha=0.3)

# Panel 1: left_stim_left_trials = stim: lick left, left stim; left_stim_right_trials = stim: lick right, left stim
plot_perf(
    axes[0],
    all_l_perf_trained,    # Control, lick left (same as all_l_perf_trained)
    left_stim_left_trials,    # Stim, lick left
    all_r_perf_trained,   # Control, lick right (same as all_r_perf_trained)
    left_stim_right_trials,   # Stim, lick right
    color_left='red',
    color_right='blue',
    title='Left Stim Side'
)

# Panel 2: right_stim_left_trials = stim: lick left, right stim; right_stim_right_trials = stim: lick right, right stim
plot_perf(
    axes[1],
    all_l_perf_trained,   # Control, lick left (same as all_l_perf_trained)
    right_stim_left_trials,   # Stim, lick left
    all_r_perf_trained,  # Control, lick right (same as all_r_perf_trained)
    right_stim_right_trials,  # Stim, lick right
    color_left='red',
    color_right='blue',
    title='Right Stim Side'
)

fig.tight_layout()
plt.show()
```

```python
save_all = []
savepaths = []

all_opto_prop_stim_left, all_opto_prop_stim_right = [],[]
all_perf_left_stim, all_perf_right_stim = [],[]
for paths in trained_paths:
    # all_opto_prop_stim_left, all_opto_prop_stim_right = [],[]

    for path in paths:
        s1 = Session(path, passive=False, filter_low_perf=True, filter_by_stim=True)
        # s1.filter_low_performance(threshold=0.550, consec_len=20)
        epoch = (s1.delay+1, s1.response)
        # epoch = (s1.sample, s1.delay)
        left_info, right_info = s1.selectivity_optogenetics(epoch=epoch,
                                                            p=0.01, 
                                                            # p=0.05/len(s1.good_neurons), 
                                                            binsize=400, 
                                                            timestep=100,
                                                            return_traces=True,
                                                            bootstrap=True)
        # sel, selo_stimleft, selo_stimright, err, erro_stimleft, erro_stimright, time = left_info
        # sel_R, selo_stimleft_R, selo_stimright_R, err_R, erro_stimleft_R, erro_stimright_R, time = right_info
        
        control_sel, opto_sel_stim_left, opto_sel_stim_right, time = left_info
        control_sel_R, opto_sel_stim_left_R, opto_sel_stim_right_R, _ = right_info
    
        period = np.where((time > s1.delay+0.1) & (time < s1.delay + 0.9))[0] # Coupling
        
        period = np.where((time > s1.delay + 1.5) & (time < s1.delay + 3))[0] # Robustness
        
        early_delay = np.where((time > s1.delay) & (time < s1.delay + 1))[0]
        late_delay = np.where((time > s1.delay + 2) & (time < s1.response))[0]
        
        # Filtering criteria
        if len(control_sel_R) > 5 and len(control_sel) > 5:
            if np.mean(np.mean(control_sel_R, axis=0)[early_delay]) >= 0.3 and np.mean(np.mean(control_sel, axis=0)[early_delay]) >= 0.3:
                if np.mean(np.mean(control_sel_R, axis=0)[late_delay]) >= 1 and np.mean(np.mean(control_sel, axis=0)[late_delay]) >= 1:
        
                    selo_stimright = np.mean(opto_sel_stim_right, axis=0)
                    selo_stimright_R = np.mean(opto_sel_stim_right_R, axis=0)
                    sel = np.mean(control_sel, axis=0)
                    selo_stimleft_R = np.mean(opto_sel_stim_left_R, axis=0)
                    selo_stimleft = np.mean(opto_sel_stim_left, axis=0)
                    sel_R = np.mean(control_sel_R, axis=0)
                    
                    # Use both hemispheres for robustness calculations

                    all_opto_prop_stim_right += [np.mean([np.mean(selo_stimleft_R[period]) / np.mean(sel_R[period]),
                                                        np.mean(selo_stimleft[period]) / np.mean(sel[period])])]

                    all_opto_prop_stim_left += [np.mean([np.mean(selo_stimright_R[period]) / np.mean(sel_R[period]), 
                                                          np.mean(selo_stimright[period]) / np.mean(sel[period])])]

                    # all_opto_prop_stim_left += [np.mean(selo_stimright[period]) / np.mean(sel[period])]
                    # all_opto_prop_stim_right += [np.mean(selo_stimleft_R[period]) / np.mean(sel_R[period])]


                    save_all += [[all_opto_prop_stim_left, all_opto_prop_stim_right]]
                    savepaths += [path]

                    _, _,control_perf = s1.performance_in_trials(s1.i_good_non_stim_trials)
                    
                    l_stim_trials = [i for i in s1.i_good_stim_trials if s1.stim_side[i] == 'L']
                    _,_,left_stim_perf = s1.performance_in_trials(l_stim_trials)
                    all_perf_left_stim += [left_stim_perf/control_perf]

                    r_stim_trials = [i for i in s1.i_good_stim_trials if s1.stim_side[i] == 'R']
                    _,_,right_stim_perf = s1.performance_in_trials(r_stim_trials)
                    all_perf_right_stim += [right_stim_perf/control_perf]

# Filtering steps 
    
all_opto_prop_stim_left, all_opto_prop_stim_right = np.array(all_opto_prop_stim_left), np.array(all_opto_prop_stim_right)
all_opto_prop_stim_left[all_opto_prop_stim_left > 1] = 1
all_opto_prop_stim_right[all_opto_prop_stim_right > 1] = 1
all_opto_prop_stim_left[all_opto_prop_stim_left < 0] = 0
all_opto_prop_stim_right[all_opto_prop_stim_right < 0] = 0
    
all_perf_left_stim = np.array(all_perf_left_stim)
all_perf_right_stim = np.array(all_perf_right_stim)
all_perf_left_stim[all_perf_left_stim > 1] = 1
all_perf_right_stim[all_perf_right_stim > 1] = 1



    # plt.scatter(all_opto_prop_stim_left, all_opto_prop_stim_right, label = path)
# plt.plot([0,1], [0,1], ls='--', color='black')
# plt.ylabel('Robustness, Right ALM')
# plt.xlabel('Robustness, Left ALM')
# # plt.legend()
# plt.show()
```
